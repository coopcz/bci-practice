```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.model_selection import cross_val_score
from mne.decoding import CSP
import numpy as np
X = epochs.get_data()
y = epochs.events[:, 2]
csp = CSP(n_components=4, reg=None, log=True, norm_trace=False)

lda = LinearDiscriminantAnalysis()
from sklearn.pipeline import Pipeline
pipeline = Pipeline([('CSP', csp), ('LDA', lda)])
scores = cross_val_score(pipeline, X, y, cv=5, scoring='accuracy')

print(f"Per-fold accuracies: {scores}")
print(f"Mean accuracy: {scores.mean():.3f} (+/- {scores.std():.3f})")
print(f"Chance level: 0.25 (4 classes)")
```

```python
montage = mne.channels.make_standard_montage('standard_1020')

channel_rename = {
    'EEG-Fz': 'Fz',
    'EEG-0': 'FC3',
    'EEG-1': 'FC1',
    'EEG-2': 'FCz',
    'EEG-3': 'FC2',
    'EEG-4': 'FC4',
    'EEG-5': 'C5',
    'EEG-C3': 'C3',
    'EEG-6': 'C1',
    'EEG-Cz': 'Cz',
    'EEG-7': 'C2',
    'EEG-C4': 'C4',
    'EEG-8': 'C6',
    'EEG-9': 'CP3',
    'EEG-10': 'CP1',
    'EEG-11': 'CPz',
    'EEG-12': 'CP2',
    'EEG-13': 'CP4',
    'EEG-14': 'P1',
    'EEG-Pz': 'Pz',
    'EEG-15': 'P2',
    'EEG-16': 'POz',
}

epochs_renamed = epochs.copy().rename_channels(channel_rename)

epochs_renamed.drop_channels(['EOG-left', 'EOG-central', 'EOG-right'])

epochs_renamed.set_montage(montage)

X_renamed = epochs_renamed.get_data()
csp.fit(X_renamed, y)

fig, axes = plt.subplots(1, 4, figsize=(12, 3))
for i, ax in enumerate(axes):
    mne.viz.plot_topomap(
        csp.filters_[i],
        epochs_renamed.info,
        axes=ax,
        show=False
    )
    ax.set_title(f'CSP Component {i+1}')

plt.suptitle('CSP Spatial Filters — Motor Imagery')
plt.tight_layout()
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from mne.decoding import CSP
from sklearn.model_selection import cross_val_score
import pandas as pd
import mne

data_path = r"C:\Users\cazie\Downloads\BCICIV_2a_gdf"

results = []

for subject_num in range(1, 10):
    filename = f"A0{subject_num}T.gdf"
    filepath = f"{data_path}\\{filename}"
    
    print(f"Processing subject {subject_num}...")
    
    # Load
    raw = mne.io.read_raw_gdf(filepath, preload=True, verbose=False)
    
    # Filter
    raw.filter(l_freq=0.5, h_freq=40.0, verbose=False)
    
    # Events — read dynamically from each file
    events, event_id = mne.events_from_annotations(raw, verbose=False)
    
    # Find the 4 motor imagery classes dynamically
    # In this dataset they're always labeled 769, 770, 771, 772
    # but MNE assigns them integer IDs that vary per file
    # So we look them up from the event_id dict directly
    mi_keys = ['769', '770', '771', '772']
    
    # Check which keys exist in this file
    available = {k: event_id[k] for k in mi_keys 
                 if k in event_id}
    
    if len(available) < 4:
        print(f"  Subject {subject_num}: missing some classes, skipping")
        print(f"  Available events: {event_id}")
        continue
    
    motor_imagery_events = {
        'left_hand': available['769'],
        'right_hand': available['770'],
        'feet': available['771'],
        'tongue': available['772']
    }
    
    # Epochs
    epochs = mne.Epochs(
        raw, events,
        event_id=motor_imagery_events,
        tmin=0.0, tmax=4.0,
        baseline=None, preload=True,
        verbose=False
    )
    
    X = epochs.get_data()
    y = epochs.events[:, 2]
    
    # CSP + LDA
    pipeline = Pipeline([
        ('CSP', CSP(n_components=4, reg=None, log=True, norm_trace=False)),
        ('LDA', LinearDiscriminantAnalysis())
    ])
    
    scores = cross_val_score(pipeline, X, y, cv=5, scoring='accuracy')
    
    results.append({
        'Subject': f'S{subject_num:02d}',
        'Mean Accuracy': round(scores.mean(), 3),
        'Std': round(scores.std(), 3),
        'Min Fold': round(scores.min(), 3),
        'Max Fold': round(scores.max(), 3)
    })
    
    print(f"  Subject {subject_num}: {scores.mean():.3f} (+/- {scores.std():.3f})")

# Build results table
df = pd.DataFrame(results)

summary = {
    'Subject': 'MEAN',
    'Mean Accuracy': round(df['Mean Accuracy'].mean(), 3),
    'Std': round(df['Std'].mean(), 3),
    'Min Fold': round(df['Min Fold'].mean(), 3),
    'Max Fold': round(df['Max Fold'].mean(), 3)
}
df = pd.concat([df, pd.DataFrame([summary])], ignore_index=True)

print("\n=== RESULTS TABLE ===")
print(df.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import numpy as np

subjects = df[df['Subject'] != 'MEAN']['Subject'].tolist()
accuracies = df[df['Subject'] != 'MEAN']['Mean Accuracy'].tolist()
stds = df[df['Subject'] != 'MEAN']['Std'].tolist()

fig, ax = plt.subplots(figsize=(10, 5))

colors = ['#2ecc71' if a >= 0.6 else '#e74c3c' if a < 0.45 
          else '#f39c12' for a in accuracies]

bars = ax.bar(subjects, accuracies, yerr=stds, 
              capsize=5, color=colors, alpha=0.85, 
              edgecolor='black', linewidth=0.8)

ax.axhline(y=0.25, color='red', linestyle='--', 
           linewidth=1.5, label='Chance level (25%)')
ax.axhline(y=df[df['Subject']=='MEAN']['Mean Accuracy'].values[0], 
           color='blue', linestyle='--', 
           linewidth=1.5, label=f'Mean accuracy (55.6%)')

ax.set_xlabel('Subject', fontsize=12)
ax.set_ylabel('Classification Accuracy', fontsize=12)
ax.set_title('CSP + LDA Motor Imagery Classification\nBCI Competition IV Dataset 2a', 
             fontsize=13)
ax.set_ylim(0, 1.0)
ax.legend(fontsize=11)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.0%}'))

for bar, acc in zip(bars, accuracies):
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.02,
            f'{acc:.1%}', ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig('csp_lda_results.png', dpi=150, bbox_inches='tight')
plt.show()

print("Figure saved as csp_lda_results.png")
```
